```python
import numpy as np # linear algebra
import pandas as pd # data processing, CSV file I/O (e.g. pd.read_csv)
import os
import seaborn as sns
import matplotlib.pyplot as plt
for dirname, _, filenames in os.walk('/kaggle/input'):
    for filename in filenames:
        print(os.path.join(dirname, filename))
import warnings
warnings.filterwarnings("ignore")
```

```python
# Carregando os dados
dados = pd.read_csv("/kaggle/input/lung-capacity-of-kids/LungCapData.csv")
```

```python
#Apresentando informações gerais sobre o conjunto de dados
dados.info()
```

```python
# Apresentando as primeiras 5 linhas
dados.head()
```

```python
# Pegando os nomes das colunas
dados.columns
```

```python
# Me livrando de uma coluna "índice"
dados.drop(columns=['Unnamed: 0'], inplace=True)
```

```python
# Informações estatísticas sobre o conjunto
dados.describe()
```

```python
# Apresentação da dispersão dos dados
plt.figure(figsize=(10, 6))
sns.scatterplot(data=dados)
plt.title("Dados")
plt.show()
```

```python
# Todos os histogramas com a frequência de cada coluna
dados.hist(edgecolor='black', linewidth=1.2)

fig = plt.gcf()  # Get current figure
fig.set_size_inches(12, 6)

plt.tight_layout()
plt.show()
```

```python
# Boxplot para identificar a presença de outliers (visualmente)
plt.figure(figsize=(12, 6))
sns.boxplot(data=dados)
plt.title('Boxplot para detecção de outliers')
plt.xticks(rotation=45)
plt.grid(True)
plt.show()
```

```python
# Matriz de correlação

plt.figure(figsize=(10, 6))
correlation = dados.corr(numeric_only=True)
sns.heatmap(correlation, annot=True, cmap='coolwarm')
plt.title("Matriz de Correlação")
plt.show()
```
